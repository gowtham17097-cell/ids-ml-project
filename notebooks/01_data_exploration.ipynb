```python
import pandas as pd
import os

# Build an absolute path to the project root, then to the dataset
# Adjust this if your notebook isn't exactly 1 folder deep

project_root = os.path.abspath(os.path.join(os.getcwd(), ".."))
data_path = os.path.join(project_root, "data", "datasets", "KDDTrain+.csv")

print("Looking for file at:", data_path)
print("File exists:", os.path.exists(data_path))

cols = [
    "duration","protocol_type","service","flag","src_bytes","dst_bytes",
    "land","wrong_fragment","urgent","hot","num_failed_logins","logged_in",
    "num_compromised","root_shell","su_attempted","num_root","num_file_creations",
    "num_shells","num_access_files","num_outbound_cmds","is_host_login",
    "is_guest_login","count","srv_count","serror_rate","srv_serror_rate",
    "rerror_rate","srv_rerror_rate","same_srv_rate","diff_srv_rate",
    "srv_diff_host_rate","dst_host_count","dst_host_srv_count",
    "dst_host_same_srv_rate","dst_host_diff_srv_rate","dst_host_same_src_port_rate",
    "dst_host_srv_diff_host_rate","dst_host_serror_rate","dst_host_srv_serror_rate",
    "dst_host_rerror_rate","dst_host_srv_rerror_rate","label","difficulty"
]

df = pd.read_csv(data_path, names=cols)

print(df.shape)
print(df["label"].value_counts())
df.head()

# Convert multi-class labels into binary: normal vs attack
df["binary_label"] = df["label"].apply(lambda x: "normal" if x == "normal" else "attack")
print(df["binary_label"].value_counts())

from sklearn.preprocessing import LabelEncoder

# Drop the multi-class label and difficulty score — we'll predict binary_label
df_model = df.drop(columns=["label", "difficulty"])

# One-hot encode categorical columns
df_model = pd.get_dummies(df_model, columns=["protocol_type", "service", "flag"])

# Encode target: normal=0, attack=1
le = LabelEncoder()
df_model["binary_label"] = le.fit_transform(df_model["binary_label"])  # normal=0, attack=1 (alphabetical)

print(df_model.shape)
df_model.head()
```
